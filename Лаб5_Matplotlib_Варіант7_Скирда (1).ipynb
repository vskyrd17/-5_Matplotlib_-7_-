### Система лінійних рівнянь (Завдання 2.2)
$$\begin{cases}
L_1: 3.2x_{1}+8.8x_{2}=89 \\
L_2: -14.2x_{1}+10.8x_{2}=125 \\
L_3: 20.3x_{1}-7.2x_{2}=142
\end{cases}$$

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Розрахунок варіанта (Завдання 2.1)
X = 'V'  # Перша велика літера імені в латинській транскрипції
N = ord(X) // 10 + 1
print(f"Розрахунок варіанта: N = ord('{X}')/10 + 1 = {N}")

# Визначення функцій ліній x2 = f(x1) для побудови графіків
def x2_L1(x1):
    """L1: 3.2*x1 + 8.8*x2 = 89"""
    return (89 - 3.2 * x1) / 8.8

def x2_L2(x1):
    """L2: -14.2*x1 + 10.8*x2 = 125"""
    return (125 + 14.2 * x1) / 10.8

def x2_L3(x1):
    """L3: 20.3*x1 - 7.2*x2 = 142"""
    return (142 - 20.3 * x1) / (-7.2)

# Визначаємо діапазон x1 для побудови. Ми визначимо його точніше в наступному кроці.
x1_range = np.linspace(-5, 25, 400)
```

```python

# --- Знаходження точок перетину (Завдання 4) ---

# 1. Точка перетину L1 та L2
A_12 = np.array([[3.2, 8.8], [-14.2, 10.8]])
B_12 = np.array([89, 125])
P_12 = np.linalg.solve(A_12, B_12)

# 2. Точка перетину L1 та L3
A_13 = np.array([[3.2, 8.8], [20.3, -7.2]])
B_13 = np.array([89, 142])
P_13 = np.linalg.solve(A_13, B_13)

# 3. Точка перетину L2 та L3
A_23 = np.array([[-14.2, 10.8], [20.3, -7.2]])
B_23 = np.array([125, 142])
P_23 = np.linalg.solve(A_23, B_23)

print("Точки перетину:")
print(f"P(L1 ∩ L2): {P_12}")
print(f"P(L1 ∩ L3): {P_13}")
print(f"P(L2 ∩ L3): {P_23}")
```

## 6. Висновки про відмінності форматів рисунків

[cite_start]У ході роботи рисунки були збережені в трьох форматах: `.jpg`, `.png` та `.svg`[cite: 22].

* **JPEG (.jpg):** Це формат із втратою даних. Файл має найменший розмір, але при сильному стисненні або масштабуванні можуть з'являтися артефакти, оскільки він оптимізований для фотографій, а не для чіткої лінійної графіки.
* [cite_start]**PNG (.png):** Це растровий формат без втрати даних[cite: 23]. Він зберігає високу якість графіки та підтримує прозорість, але розмір файлу зазвичай більший, ніж у JPEG.
* [cite_start]**SVG (.svg):** Це **векторний** формат[cite: 23].
    * **Розмір:** Для простої лінійної графіки SVG-файл зазвичай найменший, оскільки він зберігає вміст як математичні команди та опис об'єктів (векторів), а не сітку пікселів.
    * **Якість/Артефакти:** Повністю відсутні артефакти, оскільки векторна графіка масштабується без втрати якості на будь-якому збільшенні.
    * **Вміст файлу:** Файл містить XML-код, що описує лінії, кольори та текст, а не піксельні дані. Це робить його ідеальним для графіків Matplotlib, які складаються з геометричних об'єктів.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- КОМІРКА 5: ОБ'ЄДНАНИЙ ГРАФІК (Фінал) ---

# 1. Зменшення розміру рисунка (наприклад, 8x8 дюймів)
plt.figure(figsize=(8, 8), dpi=100)

# 2. Побудова ліній
plt.plot(x1_range, x2_L1_val, color='darkred', linestyle='--', linewidth=2, label='$L_1: 3.2x_{1}+8.8x_{2}=89$')
plt.plot(x1_range, x2_L2_val, color='darkgreen', linestyle='-.', linewidth=2, label='$L_2: -14.2x_{1}+10.8x_{2}=125$')
plt.plot(x1_range, x2_L3_val, color='purple', linestyle=':', linewidth=2, label='$L_3: 20.3x_{1}-7.2x_{2}=142$')

# 3. Визначення автоматичного масштабу
all_points = np.array([P_12, P_13, P_23])
x_min = np.min(all_points[:, 0]) - 5
x_max = np.max(all_points[:, 0]) + 5
y_min = np.min(all_points[:, 1]) - 5
y_max = np.max(all_points[:, 1]) + 5

# 4. Центрування осей X та Y у (0, 0)
ax = plt.gca() # Отримати поточну вісь

# Переміщення осі Y до X=0
ax.spines['left'].set_position('zero')
# Переміщення осі X до Y=0
ax.spines['bottom'].set_position('zero')
# Приховати стандартні верхню та праву рамки
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
# Додати стрілки на кінцях осей (за бажанням, але для математичного вигляду корисно)
ax.plot(1, 0, ">k", transform=ax.get_yaxis_transform(), clip_on=False)
ax.plot(0, 1, "^k", transform=ax.get_xaxis_transform(), clip_on=False)


# 5. Застосування меж
plt.xlim(x_min, x_max) # Підібрати масштаб
plt.ylim(y_min, y_max) # Підібрати масштаб

# 6. Заповнення області перетину
intersection_points = np.array([P_12, P_13, P_23])
plt.fill(intersection_points[:, 0], intersection_points[:, 1], color='orange', alpha=0.4, label='Область перетину')

# 7. Відмітка точок перетину та відповідні вказівки
plt.plot(P_12[0], P_12[1], 'o', color='blue', markersize=8)
plt.annotate(f'$P_{{12}}$: ({P_12[0]:.2f}, {P_12[1]:.2f})', P_12,
             textcoords="offset points", xytext=(-20, 20),
             arrowprops=dict(facecolor='blue', shrink=0.05))

plt.plot(P_13[0], P_13[1], 'o', color='red', markersize=8)
plt.annotate(f'$P_{{13}}$: ({P_13[0]:.2f}, {P_13[1]:.2f})', P_13,
             textcoords="offset points", xytext=(-20, -30),
             arrowprops=dict(facecolor='red', shrink=0.05))

plt.plot(P_23[0], P_23[1], 'o', color='cyan', markersize=8)
plt.annotate(f'$P_{{23}}$: ({P_23[0]:.2f}, {P_23[1]:.2f})', P_23,
             textcoords="offset points", xytext=(-20, 20),
             arrowprops=dict(facecolor='cyan', shrink=0.05))

# 8. Підписи осей та назва графіку
plt.xlabel('$X_{1}$', fontsize=14, loc='right') # Перемістити X1 на кінець осі
plt.ylabel('$X_{2}$', fontsize=14, loc='top') # Перемістити X2 на кінець осі
plt.title('Система лінійних рівнянь (Варіант 7) - Центровані осі', fontsize=16)

# 9. Додавання легенди та сітки
plt.legend(loc='upper right', fontsize=10)
plt.grid(True, color='gray', linestyle='--')

# 10. Підписи рівнянь вздовж лінії
slope_L1 = -3.2/8.8
slope_L2 = 14.2/10.8
slope_L3 = 20.3/7.2

plt.text(18, x2_L1(18)-2, '$3.2x_{1}+8.8x_{2}=89$', rotation=np.degrees(np.arctan(slope_L1)), color='darkred', fontsize=10)
plt.text(5, x2_L2(5)+2, '$-14.2x_{1}+10.8x_{2}=125$', rotation=np.degrees(np.arctan(slope_L2)), color='darkgreen', fontsize=10)
plt.text(8, x2_L3(8), '$20.3x_{1}-7.2x_{2}=142$', rotation=np.degrees(np.arctan(slope_L3)), color='purple', fontsize=10)

plt.tight_layout()
plt.show()

# --- Збереження рисунків (Завдання 5) ---
plt.savefig('combined_plot_variant_7_centered.jpg', dpi=100)
plt.savefig('combined_plot_variant_7_centered.png', dpi=100)
plt.savefig('combined_plot_variant_7_centered.svg')
print("\nГрафіки збережено у форматах .jpg, .png, .svg")
```
